## Pandas + NumPy Quick Revision

### Series vs DataFrame
- Series → 1D labeled structure, usually one column
- DataFrame → 2D table with rows and columns

### DataFrame Shape

`df.shape` returns:

(rows, columns)

Example:

(1000, 8) → 1000 rows and 8 columns

### Vectorization

Pandas and NumPy can perform operations on entire columns/arrays
without manually writing Python loops.

Example:

df["Revenue"] = df["Price"] * df["Quantity"]

### Missing Values

df.isnull().sum()

→ Number of missing values in each column.

`dropna()` → removes missing observations (rows by default).

`fillna()` → replaces missing values.

Do not automatically drop or fill missing data.
The decision depends on the data and analytical problem.

### value_counts()

Counts the frequency of each unique value.

### NumPy

NumPy is used for efficient numerical computing.

Important:
- Fast numerical operations
- Vectorized operations
- Boolean filtering
- Multidimensional arrays

### Boolean Filtering

scores[scores >= 70]

Condition creates a Boolean mask and returns values
where the condition is True.

### Array Dimensions

A NumPy array is not automatically a matrix.

np.array([1, 2, 3]) → 1D

np.array([[1, 2], [3, 4]]) → 2D

Use:

arr.ndim

to check the number of dimensions.
